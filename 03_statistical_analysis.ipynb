```python
import pandas as pd
import numpy as np
import scipy.stats as stats
import matplotlib.pyplot as plt
import seaborn as sns
import statsmodels.api as sm
```

```python
# Read in cleaned data csv

data = pd.read_csv('data/clean/cleaned_data.csv')
```

# Hypothesis Test with Logistic Regression in Statsmodels

##### Below, we will run a statistical test to determine whether *Job Level* and *Monthly Income* have statistically significant relationships with **Attrition** and whether the results support our hypothesis. As previously stated, our hypothesis is that employees with lower *Compensation* and lower *Job Levels* are more likely to leave the organization than employees with higher *Compensation* and higher *Job Levels*.

```python
# Run Statistical Analysis with Logistic Regression with Stats Model

X = data[['MonthlyIncome', 'JobLevel']]
y = data['Attrition']

X = sm.add_constant(X)

model = sm.Logit(y, X)
results = model.fit()

print(results.summary())
```

# Logistic Regression Analysis and Statistical Significance Conclusion

### Now that we have the results, we can see that with our p-values being higher than our alpha of 0.05, we failed to reject the null hypothesis in this analysis. Both Monthly Income (p-value: 0.549) and Job Level (p-value: 0.061) did not show a statistically significant relationship with Attrition when analyzed together. With these results we will now move onto the model creation phase.
