```python
# Import all libraries that will be needed for this capstone project for the exploratory data analysis

import pandas as pd
import numpy as np

pd.set_option('display.max_columns', None)
```

```python
# Read in data and look at first 5 columns of the dataset
df = pd.read_csv('data/raw/Employee-Attrition.csv')
data = df.copy()
data.head()

# help(pd.read_csv)
```

```python
# Analyze data with .info() and check total memory usage
data.info()
```

```python
# Checking Missing values in the dataset by print a the unique value counts of the amount of missing values per column 
missing_values = data.isnull().sum().value_counts()
missing_values
```

```python
data.head()
```

```python
# Find all object columns and convert column names to a list
cats = data.select_dtypes('object').columns.to_list()
cats
```

```python
# Loop through cats list and print unique value counts per column to inspect cardinality

for cat in cats:
    print(f'Unique Counts for {cat}: {data[cat].value_counts()}\n')
```

```python
data['Attrition'] = data['Attrition'].map({'Yes': 1, 'No': 0})
data['OverTime'] = data['OverTime'].map({'Yes': 1, 'No': 0})
data['BusinessTravel'] = data['BusinessTravel'].map({'Non-Travel': 0, 'Travel_Rarely': 1, 'Travel_Frequently': 2})
data.head(15)
```

```python
data['Attrition'].mean()
```

```python
data[['StandardHours', 'Over18', 'EmployeeCount']].value_counts()
```

```python
data = data.drop(columns=['StandardHours', 'Over18', 'EmployeeCount'])
```

```python
# Convert all string type columns to category to help save memory and performance 

cats = data.select_dtypes('object').columns.to_list()
data[cats] = data[cats].astype('category')
data.select_dtypes('object').columns.to_list()
data.info()
```

```python
data.to_csv('data/clean/cleaned_data.csv', index=False)
```
